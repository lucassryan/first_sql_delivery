```sql
-- Exercício 1
-- Liste todos os clientes que moram na cidade de Uberlândia.
SELECT 
  * 
FROM 
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  cidade = 'Uberlândia'
```

```sql
-- Exercício 2
-- Liste os clientes cujo bairro seja 'Centro'.
SELECT
  id_cliente,
  email
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  bairro = 'Centro'
```

```sql
-- Exercício 3
-- Liste os clientes com consumo_mensal_kwh maior que 3000.
SELECT
  email,
  consumo_mensal_kwh
  FROM
    primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
  WHERE
    consumo_mensal_kwh > 3000
```

```sql
-- Exercício 4
-- Liste os clientes com consumo_mensal_kwh menor que 200.
SELECT
  email,
  consumo_mensal_kwh
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  consumo_mensal_kwh <200
```

```sql
-- Exercício 5
-- Liste os clientes cuja classe_consumidor seja 'Rural'.
SELECT
  *
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  classe_consumidor = 'Rural'
```

```sql
-- Exercício 6
-- Liste os clientes do mercado 'Cativo' com consumo_mensal_kwh maior que 1500.
SELECT
  email,
  id_cliente,
  consumo_mensal_kwh
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  tipo_mercado = 'Cativo'
AND
  consumo_mensal_kwh > 1500
```

```sql
-- Exercício 7 
-- Esse exercício o chat fez errado, por que ele pede apenas os que terminam com (.com)
-- pesquisei e vi que tem como usar o RLIKE, mas pra primeira semana vou manter assim.
-- Liste os clientes cujo email termine com '.com'.
SELECT
  id_cliente,
  email
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  email LIKE '%.com%'
```

```sql
-- Exercício 8
-- Liste os clientes cujo email NÃO contenha 'hotmail'.
SELECT
  id_cliente,
  email
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  email
NOT LIKE
  '%hotmail%'
```

```sql
-- Exercício 9
-- Liste os clientes que moram em Betim, Contagem ou Juiz de Fora.
SELECT
  id_cliente,
  email,
  cidade
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  cidade IN ('Betim', 'Contagem', 'Juiz de Fora')
```

```sql
-- Exercício 10
-- Liste os clientes que NÃO moram em Betim, Contagem ou Juiz de Fora.
SELECT
  email,
  id_cliente,
  cidade
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  cidade
NOT LIKE IN ('Betim', 'Contagem', 'Juiz de Fora')
```

```sql
-- Exercício 11
-- Liste os clientes ordenados por consumo_mensal_kwh em ordem crescente.
SELECT
  id_cliente,
  email,
  consumo_mensal_kwh
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  consumo_mensal_kwh ASC
```

```sql
-- Exercício 12
-- Liste os clientes ordenados por consumo_mensal_kwh em ordem decrescente.
SELECT
  id_cliente,
  email,
  consumo_mensal_kwh
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  consumo_mensal_kwh DESC
```

```sql
-- Exercício 13
-- Liste os clientes ordenados por cidade em ordem alfabética (A–Z).
SELECT
  id_cliente,
  email,
  cidade
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  cidade ASC
```

```sql
-- Exercício 14
-- Liste os clientes ordenados por mes_referencia do mais antigo para o mais recente.
SELECT
  id_cliente,
  email,
  mes_referencia
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  mes_referencia ASC
```

```sql
-- Exercício 15
-- Liste os 5 clientes com maior consumo_mensal_kwh.
SELECT
  id_cliente,
  email,
  consumo_mensal_kwh
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  consumo_mensal_kwh DESC
LIMIT 5
```

```sql
-- Exercício 16
-- Liste os 5 clientes com menor consumo_mensal_kwh.
SELECT
  id_cliente,
  email,
  consumo_mensal_kwh
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  consumo_mensal_kwh ASC
LIMIT 5
```

```sql
-- Exercício 17
-- Liste os 10 primeiros clientes ordenando por id_cliente em ordem crescente.
SELECT
  id_cliente,
  email
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  id_cliente ASC
LIMIT 10
```

```sql
-- Exercício 18
-- Liste os 10 clientes seguintes, pulando os 10 primeiros,
-- ordenando por id_cliente em ordem crescente.
SELECT
  id_cliente,
  email
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
ORDER BY
  id_cliente ASC
LIMIT 10 OFFSET 10
```

```sql
-- Exercício 19
-- Liste os clientes da classe 'Comercial' que sejam do mercado 'Livre'.
SELECT
  id_cliente,
  email,
  classe_consumidor,
  tipo_mercado
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  classe_consumidor = 'Comercial'
AND
  tipo_mercado = 'Livre'
```

```sql
-- Exercício 20
-- Liste os clientes que moram em Belo Horizonte,
-- tenham consumo_mensal_kwh maior que 5000
-- e sejam do mercado 'Livre'.
SELECT
  id_cliente,
  email,
  cidade,
  consumo_mensal_kwh,
  tipo_mercado
FROM
  primeira_entrega.setor_eletrico.clientes_setor_eletrico_mg
WHERE
  cidade = 'Belo Horizonte' AND consumo_mensal_kwh < 5000 AND tipo_mercado = 'Livre'
```
